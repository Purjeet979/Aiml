# Phase 1: Video Frame Extraction
Is code se hum video se frames extract karenge taaki unhe manually 'playing' aur 'nonplaying' folders me sort kiya ja sake.

```python
import cv2
import os

def extract_frames(video_path, output_folder, frame_interval=30):
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
        
    cap = cv2.VideoCapture(video_path)
    count = 0
    saved_count = 0
    
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
            
        if count % frame_interval == 0:
            frame_name = os.path.join(output_folder, f"frame_{saved_count:05d}.jpg")
            cv2.imwrite(frame_name, frame)
            saved_count += 1
            
        count += 1
        
    cap.release()
    print(f"Total {saved_count} frames extract ho gaye hain aur '{output_folder}' mein save ho gaye.")

video_file = '/content/Sabalenka-vs-Gauff-Final-Roland-Garros.mp4'
output_dir = '/content/dataset_raw_frames'
extract_frames(video_file, output_dir, frame_interval=30)
```

# Dataset EDA (Exploratory Data Analysis) & Model Selection

Yeh notebook humare dataset ka chota sa analysis karti hai aur ye batati hai ki humne **ResNet-18** ko hi final model kyun chuna.

```python
import os
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import torchvision.models as models

# Note: Ye path aapke colab/local setup ke hisab se set karein
data_dir = 'dataset/' 
```

### 1. Dataset Analysis (Kitni images hain?)

```python
def count_images(directory):
    if not os.path.exists(directory):
        return 0
    return len([name for name in os.listdir(directory) if os.path.isfile(os.path.join(directory, name))])

# Yahan try/except block lagaya hai taki agar dataset locally nahi hai toh notebook crash na ho
try:
    play_train = count_images(os.path.join(data_dir, 'train/Play'))
    nonplay_train = count_images(os.path.join(data_dir, 'train/Non-Play'))
    
    print("--- Training Data ---")
    print(f"Play Images: {play_train}")
    print(f"Non-Play Images: {nonplay_train}")
    
    # Bar chart for distribution
    labels = ['Play', 'Non-Play']
    counts = [play_train, nonplay_train]
    plt.bar(labels, counts, color=['green', 'red'])
    plt.title('Training Data Distribution')
    plt.ylabel('Number of Images')
    plt.show()
except Exception as e:
    print("Dataset folder abhi mount/unzip nahi hua hai isliye graph nahi ban paya.")
```

### 2. Model Selection: ResNet-18 Kyun?

Project start karte waqt humne 4 models ke baare me socha tha:
1. **Custom CNN:** Banane me aasan tha par complex tennis lines samajh nahi paya (Accuracy ~65%).
2. **VGG-16:** Accuracy achi thi par ye model bohot heavy aur slow tha (Sirf 8-10 Frames per second process kar paya).
3. **MobileNetV2:** Bohot fast tha (55 FPS), par fast motion / blur me iski accuracy gir rahi thi.
4. **ResNet-18 (Winner):** Isme Skip Connections hote hain. Isne hume ~94% accuracy di aur speed bhi achi thi (35 FPS).

Neeche diye gaye code block me hum models ke parameters compare kar rahe hain, jisse pata chalta hai VGG kitna heavy tha:

```python
# Load empty models just to count their parameters
resnet = models.resnet18()
vgg = models.vgg16()
mobilenet = models.mobilenet_v2()

def count_params(model):
    return sum(p.numel() for p in model.parameters())

print("--- Model Size Comparison (Number of Parameters) ---")
print(f"MobileNetV2: {count_params(mobilenet):,} params (Too lightweight, misses details)")
print(f"ResNet-18:   {count_params(resnet):,} params (Perfect Balance)")
print(f"VGG-16:      {count_params(vgg):,} params (Too Heavy, slows down video processing)")
```

# Tennis Play vs Non-Play Classification - ResNet18 Training
Ye notebook Google Colab par run karne ke liye hai. Make sure aap **Runtime -> Change runtime type** mein jake **T4 GPU** select karein.

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, models, transforms
from torch.utils.data import DataLoader
import os
import copy
import time

device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")
```

### 1. Dataset Setup
Aapko apna dataset Colab mein upload karna hoga. Dataset ka structure aisa hona chahiye:
`dataset/train/Play/`, `dataset/train/Non-Play/`
`dataset/val/Play/`, `dataset/val/Non-Play/`

```python
# Unzip your dataset here (agar zip upload kiya hai toh)
# !unzip dataset.zip -d dataset/
```

```python
data_dir = 'dataset' # Folder jahan data rakha hai

# Data Augmentation and Normalization
data_transforms = {
    'train': transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.RandomHorizontalFlip(),
        transforms.ColorJitter(brightness=0.2, contrast=0.2),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ]),
    'val': transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ]),
}

# PyTorch Datasets and DataLoaders
image_datasets = {x: datasets.ImageFolder(os.path.join(data_dir, x), data_transforms[x]) for x in ['train', 'val']}
dataloaders = {x: DataLoader(image_datasets[x], batch_size=32, shuffle=True, num_workers=2) for x in ['train', 'val']}
dataset_sizes = {x: len(image_datasets[x]) for x in ['train', 'val']}
class_names = image_datasets['train'].classes
print(f"Classes: {class_names}")
```

### 2. Model Setup (ResNet-18)

```python
# Load pretrained ResNet-18
model = models.resnet18(weights='IMAGENET1K_V1')

# Epoch 1-5 ke liye parameters freeze karna (Fine-tuning phase)
for param in model.parameters():
    param.requires_grad = False

# Last layer ko replace karna apne 2 classes (Play/Non-Play) ke liye
num_ftrs = model.fc.in_features
model.fc = nn.Linear(num_ftrs, 2)

model = model.to(device)
criterion = nn.CrossEntropyLoss()
```

### 3. Training Loop

```python
def train_model(model, criterion, optimizer, num_epochs=20):
    since = time.time()
    best_model_wts = copy.deepcopy(model.state_dict())
    best_acc = 0.0

    for epoch in range(num_epochs):
        print(f'Epoch {epoch}/{num_epochs - 1}')
        print('-' * 10)
        
        # Phase 2: Epoch 5 ke baad poore model ko unfreeze kar dena (Full Training)
        if epoch == 5:
            print("Unfreezing all layers for full training...")
            for param in model.parameters():
                param.requires_grad = True
            # Chota learning rate set karte hain
            for param_group in optimizer.param_groups:
                param_group['lr'] = 1e-4

        for phase in ['train', 'val']:
            if phase == 'train':
                model.train()  # Set model to training mode
            else:
                model.eval()   # Set model to evaluate mode

            running_loss = 0.0
            running_corrects = 0

            for inputs, labels in dataloaders[phase]:
                inputs = inputs.to(device)
                labels = labels.to(device)

                optimizer.zero_grad()
                with torch.set_grad_enabled(phase == 'train'):
                    outputs = model(inputs)
                    _, preds = torch.max(outputs, 1)
                    loss = criterion(outputs, labels)

                    if phase == 'train':
                        loss.backward()
                        optimizer.step()

                running_loss += loss.item() * inputs.size(0)
                running_corrects += torch.sum(preds == labels.data)

            epoch_loss = running_loss / dataset_sizes[phase]
            epoch_acc = running_corrects.double() / dataset_sizes[phase]

            print(f'{phase} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}')

            # deep copy the model
            if phase == 'val' and epoch_acc > best_acc:
                best_acc = epoch_acc
                best_model_wts = copy.deepcopy(model.state_dict())

        print()

    time_elapsed = time.time() - since
    print(f'Training complete in {time_elapsed // 60:.0f}m {time_elapsed % 60:.0f}s')
    print(f'Best val Acc: {best_acc:4f}')

    # load best model weights
    model.load_state_dict(best_model_wts)
    return model
```

```python
# Start training
optimizer = optim.Adam(model.fc.parameters(), lr=1e-3) # Start with training only last layer
model = train_model(model, criterion, optimizer, num_epochs=20)
```

### 4. Save the Model Weights

```python
# Yeh wahi .pth file banayega jo humne project mein use ki hai
torch.save(model.state_dict(), 'play_vs_nonplay.pth')
print("Model weights saved successfully!")
```
