# Tennis Video - Custom Dataset Creation

Yeh notebook dikhata hai ki humne videos se frames (images) kaise extract kiye the apna custom dataset banane ke liye. Hum video ke har N-th frame ko save karenge (jaise har 5th ya 6th frame) taaki duplicate data kam ho.

```python
import cv2
import os
from google.colab import drive

# Google Drive mount karna (agar data Drive par hai)
drive.mount('/content/drive')
```

### Step 1: Extract Frames from Video

```python
def extract_frames(video_path, output_folder, frame_interval=5):
    """
    video_path: Video file ka path
    output_folder: Kahan frames save karne hain
    frame_interval: Har kitne frames ke baad ek frame save karna hai (e.g. 5 means har 5th frame)
    """
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
        
    cap = cv2.VideoCapture(video_path)
    count = 0
    saved_count = 0
    
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
            
        # Sirf tabhi save karein jab count frame_interval se divide ho (jaise har 5th frame)
        if count % frame_interval == 0:
            frame_name = os.path.join(output_folder, f"frame_{saved_count:05d}.jpg")
            cv2.imwrite(frame_name, frame)
            saved_count += 1
            
        count += 1
        
    cap.release()
    print(f"Total {saved_count} frames extract ho gaye hain aur '{output_folder}' mein save ho gaye.")

# Example usage:
video_file = '/content/drive/MyDrive/Tennis_Project/sample_match.mp4'  # Apna video path yahan dalein
output_dir = '/content/dataset_raw_frames'

# Extract every 6th frame (agar 30 fps video hai, toh 1 second me 5 frames milenge)
extract_frames(video_file, output_dir, frame_interval=6)
```

### Step 2: Manual Labeling Process (Play vs Non-Play)

Frames extract hone ke baad humara process aisa tha:
1. **Folder Structure:** Humne pehle do folders banaye: `Play` aur `Non-Play`.
2. **Manual Sorting (Drag & Drop):** Extract kiye gaye frames ko apne computer ya Google Drive mein open kiya.
   - Jo images mein tennis court ka wide broadcast view dikh raha tha, net aur dono players positions par the = unko `Play` folder me daal diya.
   - Jo images mein close-ups, crowd, players baithe hue, replay, ya alag angle tha = unko `Non-Play` folder me daal diya.
3. Yehi sorted folders (`Play/`, `Non-Play/`) hi aapka main training dataset ban gaya!
