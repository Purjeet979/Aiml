# Dataset EDA (Exploratory Data Analysis) & Model Selection

Yeh notebook humare dataset ka chota sa analysis karti hai aur ye batati hai ki humne **ResNet-18** ko hi final model kyun chuna.

```python
import os
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import torchvision.models as models

# Note: Ye path aapke colab/local setup ke hisab se set karein
data_dir = 'dataset/' 
```

### 1. Dataset Analysis (Kitni images hain?)

```python
def count_images(directory):
    if not os.path.exists(directory):
        return 0
    return len([name for name in os.listdir(directory) if os.path.isfile(os.path.join(directory, name))])

# Yahan try/except block lagaya hai taki agar dataset locally nahi hai toh notebook crash na ho
try:
    play_train = count_images(os.path.join(data_dir, 'train/Play'))
    nonplay_train = count_images(os.path.join(data_dir, 'train/Non-Play'))
    
    print("--- Training Data ---")
    print(f"Play Images: {play_train}")
    print(f"Non-Play Images: {nonplay_train}")
    
    # Bar chart for distribution
    labels = ['Play', 'Non-Play']
    counts = [play_train, nonplay_train]
    plt.bar(labels, counts, color=['green', 'red'])
    plt.title('Training Data Distribution')
    plt.ylabel('Number of Images')
    plt.show()
except Exception as e:
    print("Dataset folder abhi mount/unzip nahi hua hai isliye graph nahi ban paya.")
```

### 2. Model Selection: ResNet-18 Kyun?

Project start karte waqt humne 4 models ke baare me socha tha:
1. **Custom CNN:** Banane me aasan tha par complex tennis lines samajh nahi paya (Accuracy ~65%).
2. **VGG-16:** Accuracy achi thi par ye model bohot heavy aur slow tha (Sirf 8-10 Frames per second process kar paya).
3. **MobileNetV2:** Bohot fast tha (55 FPS), par fast motion / blur me iski accuracy gir rahi thi.
4. **ResNet-18 (Winner):** Isme Skip Connections hote hain. Isne hume ~94% accuracy di aur speed bhi achi thi (35 FPS).

Neeche diye gaye code block me hum models ke parameters compare kar rahe hain, jisse pata chalta hai VGG kitna heavy tha:

```python
# Load empty models just to count their parameters
resnet = models.resnet18()
vgg = models.vgg16()
mobilenet = models.mobilenet_v2()

def count_params(model):
    return sum(p.numel() for p in model.parameters())

print("--- Model Size Comparison (Number of Parameters) ---")
print(f"MobileNetV2: {count_params(mobilenet):,} params (Too lightweight, misses details)")
print(f"ResNet-18:   {count_params(resnet):,} params (Perfect Balance)")
print(f"VGG-16:      {count_params(vgg):,} params (Too Heavy, slows down video processing)")
```
